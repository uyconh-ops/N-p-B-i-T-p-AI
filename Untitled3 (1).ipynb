```python
import numpy as np
import skfuzzy as fuzz
from skfuzzy import control as ctrl

khoang_cach = ctrl.Antecedent(np.arange(0, 11, 1), 'khoang_cach')
giao_thong = ctrl.Antecedent(np.arange(0, 11, 1), 'giao_thong')
nhu_cau = ctrl.Antecedent(np.arange(0, 11, 1), 'nhu_cau')

gia = ctrl.Consequent(np.arange(0, 101, 1), 'gia')

khoang_cach['gan'] = fuzz.trimf(khoang_cach.universe, [0, 0, 5])
khoang_cach['trung_binh'] = fuzz.trimf(khoang_cach.universe, [0, 5, 10])
khoang_cach['xa'] = fuzz.trimf(khoang_cach.universe, [5, 10, 10])

giao_thong['thap'] = fuzz.trimf(giao_thong.universe, [0, 0, 5])
giao_thong['trung_binh'] = fuzz.trimf(giao_thong.universe, [0, 5, 10])
giao_thong['cao'] = fuzz.trimf(giao_thong.universe, [5, 10, 10])

nhu_cau['thap'] = fuzz.trimf(nhu_cau.universe, [0, 0, 5])
nhu_cau['trung_binh'] = fuzz.trimf(nhu_cau.universe, [0, 5, 10])
nhu_cau['cao'] = fuzz.trimf(nhu_cau.universe, [5, 10, 10])

gia['thap'] = fuzz.trimf(gia.universe, [0, 0, 50])
gia['trung_binh'] = fuzz.trimf(gia.universe, [25, 50, 75])
gia['cao'] = fuzz.trimf(gia.universe, [50, 100, 100])

r1 = ctrl.Rule(khoang_cach['gan'] & nhu_cau['thap'], gia['thap'])
r2 = ctrl.Rule(khoang_cach['xa'] & nhu_cau['cao'], gia['cao'])
r3 = ctrl.Rule(giao_thong['cao'] & nhu_cau['cao'], gia['cao'])
r4 = ctrl.Rule(giao_thong['thap'] & nhu_cau['thap'], gia['thap'])
r5 = ctrl.Rule(khoang_cach['trung_binh'], gia['trung_binh'])

system = ctrl.ControlSystem([r1,r2,r3,r4,r5])
sim = ctrl.ControlSystemSimulation(system)

sim.input['khoang_cach'] = 8
sim.input['giao_thong'] = 7
sim.input['nhu_cau'] = 9

sim.compute()
print("Giá:", sim.output['gia'])
gia.view(sim=sim)
```

```python
danh_gia = ctrl.Antecedent(np.arange(0,11,1),'danh_gia')
doanh_so = ctrl.Antecedent(np.arange(0,11,1),'doanh_so')
loi_nhuan = ctrl.Antecedent(np.arange(0,11,1),'loi_nhuan')
doi_thu = ctrl.Antecedent(np.arange(0,11,1),'doi_thu')

giam_gia = ctrl.Consequent(np.arange(0,71,1),'giam_gia')

for var in [danh_gia, doanh_so, loi_nhuan, doi_thu]:
    var['thap'] = fuzz.trimf(var.universe,[0,0,5])
    var['trung_binh'] = fuzz.trimf(var.universe,[0,5,10])
    var['cao'] = fuzz.trimf(var.universe,[5,10,10])

giam_gia['thap'] = fuzz.trimf(giam_gia.universe,[0,0,20])
giam_gia['trung_binh'] = fuzz.trimf(giam_gia.universe,[10,30,50])
giam_gia['cao'] = fuzz.trimf(giam_gia.universe,[40,70,70])

r1 = ctrl.Rule(danh_gia['cao'] & doanh_so['cao'], giam_gia['thap'])
r2 = ctrl.Rule(doi_thu['cao'] & loi_nhuan['thap'], giam_gia['cao'])
r3 = ctrl.Rule(doanh_so['trung_binh'], giam_gia['trung_binh'])
r4 = ctrl.Rule(danh_gia['thap'], giam_gia['cao'])
r5 = ctrl.Rule(loi_nhuan['cao'] & doi_thu['thap'], giam_gia['thap'])

system2 = ctrl.ControlSystem([r1,r2,r3,r4,r5])
sim2 = ctrl.ControlSystemSimulation(system2)

sim2.input['danh_gia'] = 4
sim2.input['doanh_so'] = 5
sim2.input['loi_nhuan'] = 3
sim2.input['doi_thu'] = 8

sim2.compute()
print("Giảm giá:", sim2.output['giam_gia'])
giam_gia.view(sim=sim2)
```

```python
nhu_cau = ctrl.Antecedent(np.arange(0,11,1),'nhu_cau')
canh_tranh = ctrl.Antecedent(np.arange(0,11,1),'canh_tranh')
uy_tin = ctrl.Antecedent(np.arange(0,11,1),'uy_tin')
loi_nhuan = ctrl.Antecedent(np.arange(0,11,1),'loi_nhuan')

chien_luoc = ctrl.Consequent(np.arange(0,101,1),'chien_luoc')

for var in [nhu_cau, canh_tranh, uy_tin, loi_nhuan]:
    var['thap'] = fuzz.trimf(var.universe,[0,0,5])
    var['trung_binh'] = fuzz.trimf(var.universe,[0,5,10])
    var['cao'] = fuzz.trimf(var.universe,[5,10,10])

chien_luoc['thap'] = fuzz.trimf(chien_luoc.universe,[0,0,50])
chien_luoc['trung_binh'] = fuzz.trimf(chien_luoc.universe,[25,50,75])
chien_luoc['cao'] = fuzz.trimf(chien_luoc.universe,[50,100,100])

r1 = ctrl.Rule(nhu_cau['cao'] & canh_tranh['cao'], chien_luoc['cao'])
r2 = ctrl.Rule(loi_nhuan['cao'] & canh_tranh['thap'], chien_luoc['thap'])
r3 = ctrl.Rule(uy_tin['cao'], chien_luoc['trung_binh'])
r4 = ctrl.Rule(nhu_cau['thap'], chien_luoc['thap'])

system3 = ctrl.ControlSystem([r1,r2,r3,r4])
sim3 = ctrl.ControlSystemSimulation(system3)

sim3.input['nhu_cau'] = 9
sim3.input['canh_tranh'] = 6
sim3.input['uy_tin'] = 7
sim3.input['loi_nhuan'] = 8

sim3.compute()
print("Chiến lược:", sim3.output['chien_luoc'])
chien_luoc.view(sim=sim3)
```

```python
mat_do = ctrl.Antecedent(np.arange(0,11,1),'mat_do')
khan_cap = ctrl.Antecedent(np.arange(0,11,1),'khan_cap')
tai_trong = ctrl.Antecedent(np.arange(0,11,1),'tai_trong')
giao_thong = ctrl.Antecedent(np.arange(0,11,1),'giao_thong')

uu_tien = ctrl.Consequent(np.arange(0,101,1),'uu_tien')

for var in [mat_do, khan_cap, tai_trong, giao_thong]:
    var['thap'] = fuzz.trimf(var.universe,[0,0,5])
    var['trung_binh'] = fuzz.trimf(var.universe,[0,5,10])
    var['cao'] = fuzz.trimf(var.universe,[5,10,10])

uu_tien['thap'] = fuzz.trimf(uu_tien.universe,[0,0,50])
uu_tien['trung_binh'] = fuzz.trimf(uu_tien.universe,[25,50,75])
uu_tien['cao'] = fuzz.trimf(uu_tien.universe,[50,100,100])

r1 = ctrl.Rule(khan_cap['cao'], uu_tien['cao'])
r2 = ctrl.Rule(mat_do['cao'] & tai_trong['thap'], uu_tien['cao'])
r3 = ctrl.Rule(giao_thong['cao'], uu_tien['thap'])
r4 = ctrl.Rule(khan_cap['trung_binh'], uu_tien['trung_binh'])

system4 = ctrl.ControlSystem([r1,r2,r3,r4])
sim4 = ctrl.ControlSystemSimulation(system4)

sim4.input['mat_do'] = 8
sim4.input['khan_cap'] = 7
sim4.input['tai_trong'] = 3
sim4.input['giao_thong'] = 6

sim4.compute()
print("Ưu tiên:", sim4.output['uu_tien'])
uu_tien.view(sim=sim4)
```
